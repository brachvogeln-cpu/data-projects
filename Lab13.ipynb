```python
import pandas as pd
import plotly.express as px
```

```python
fire = pd.read_csv("forestfires.csv")
fire
```

```python
firescatter = px.scatter(fire, 
                 x = "temp",
                     y = "area", 
                        title = "Temperature and Area of Forest Fires in Portugal", 
                        labels = {"area": "Area",
                                  "temp": "Temperature"})
firescatter
```
